```python
def is_palindrome(text):
    # Input Constraint: 'text' must be a string.
    # Output Constraint: Returns a boolean (True/False).

    if not isinstance(text, str):
        raise TypeError("Input must be a string.")

    # Preprocess the string: convert to lowercase and remove non-alphanumeric characters
    clean_text = "".join(char for char in text if char.isalnum()).lower()

    # Check if the clean string is equal to its reverse
    return clean_text == clean_text[::-1]
```

```python
print(f"'racecar' is palindrome: {is_palindrome('racecar')}")                 # Expected: True
print(f"'Racecar' is palindrome: {is_palindrome('Racecar')}")                 # Expected: True (case-insensitive)
print(f"'A man, a plan, a canal: Panama' is palindrome: {is_palindrome('A man, a plan, a canal: Panama')}") # Expected: True (ignore non-alphanumeric)
print(f"'hello' is palindrome: {is_palindrome('hello')}")                     # Expected: False
print(f"'madam' is palindrome: {is_palindrome('madam')}")                     # Expected: True
print(f"'12321' is palindrome: {is_palindrome('12321')}")                     # Expected: True
print(f"'No lemon, no melon' is palindrome: {is_palindrome('No lemon, no melon')}") # Expected: True
print(f"'' is palindrome: {is_palindrome('')}")                               # Expected: True (empty string)

# Edge case: Non-string input
try:
    is_palindrome(123)
except TypeError as e:
    print(f"Error for 123: {e}")
```
