```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 150

sqft = np.random.randint(800, 3500, size=n)
locations = np.random.choice(
    ["Downtown", "Suburbs", "Uptown", "Rural"], size=n
)
# Price formula: $200/sqft + location premium + noise
location_premium = {
    "Downtown": 100000,
    "Uptown": 75000,
    "Suburbs": 30000,
    "Rural": 0,
}
base_price = (
    sqft * 200
    + [location_premium[loc] for loc in locations]
    + np.random.normal(0, 15000, size=n)
)

housing_df = pd.DataFrame(
    {"square_footage": sqft, "location": locations, "price": base_price.round()}
)

# Source citation comment required by assignment:
# Data Source: Generated synthetic housing dataset (150 records) based on US real estate patterns.
housing_df.to_csv("housing_data.csv", index=False)
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder

# Data Source: Generated synthetic housing dataset (150 records)
df = pd.read_csv("housing_data.csv")

# 1. Separate features and target
X_num = df[["square_footage"]]
X_cat = df[["location"]]
y = df["price"]

# 2. One-Hot Encode categorical variable 'location'
encoder = OneHotEncoder(sparse_output=False)
X_cat_encoded = encoder.fit_transform(X_cat)
encoded_feature_names = encoder.get_feature_names_out(["location"])

# Combine features
X_combined = np.hstack([X_num.values, X_cat_encoded])

# 3. Train Model
model = LinearRegression()
model.fit(X_combined, y)

# 4. Predict price for a 2000 sq ft house in Downtown
new_house_cat = encoder.transform(pd.DataFrame([["Downtown"]], columns=["location"]))
new_house_features = np.hstack([[2000], new_house_cat[0]]).reshape(1, -1)
predicted_price = model.predict(new_house_features)

print(f"Predicted Price for 2000 sq ft in Downtown: ${predicted_price[0]:,.2f}")
print("\nModel Coefficients:")
print(f"Square Footage Coefficient: ${model.coef_[0]:.2f} per sq ft")
for name, coef in zip(encoded_feature_names, model.coef_[1:]):
  print(f"{name}: ${coef:,.2f}")
```
