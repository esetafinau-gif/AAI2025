## Exercise 1: Prompt Chaining for a Customer Support AI

This exercise demonstrates a multi-step prompt chain for a customer support AI. The goal is to build a coherent flow where the output of one prompt becomes the input for the next, guiding the AI through a customer interaction from issue classification to potential escalation.

### Step 1: Classify Customer Issue

**Goal:** Accurately identify the category and urgency of the customer's problem based on their initial statement. This output will inform subsequent steps.

**Prompt:**

```markdown
You are an experienced customer support agent. Your task is to classify a customer's issue based on their message.

**Customer Message:** {customer_message}

Analyze the message and provide:
1.  **Issue Category:** (e.g., 'Billing', 'Technical Support', 'Product Inquiry', 'Account Management', 'Other')
2.  **Urgency Level:** (e.g., 'Low', 'Medium', 'High', 'Critical')
3.  **Key Details:** A brief summary of the main points.

Format your response as a JSON object with keys 'category', 'urgency', and 'details'.
```

### Step 2: Gather Missing Information (if needed)

**Goal:** If the initial classification indicates missing information, the AI will ask clarifying questions. This step leverages the `Key Details` from the previous step.

**Prompt:**

```markdown
You are a helpful customer support agent. Based on the customer's issue and initial classification, determine if more information is needed to resolve the problem. If so, formulate a polite and concise question.

**Issue Category:** {issue_category}
**Urgency Level:** {urgency_level}
**Key Details:** {key_details}

Consider the 'Key Details'. Do you need more information?

If YES, provide a single question to the customer to get the necessary information. Start with 'To help you better, could you please...'
If NO, respond with 'No further information needed at this stage.'
```

### Step 3: Propose a Solution or Next Steps

**Goal:** Based on all gathered information, propose a solution or clear next steps for the customer. This combines outputs from Step 1 and Step 2.

**Prompt:**

```markdown
You are a diligent customer support agent. Provide a clear, actionable solution or next steps for the customer based on their classified issue and any gathered information. Maintain a supportive and professional tone.

**Issue Category:** {issue_category}
**Urgency Level:** {urgency_level}
**Key Details:** {key_details}
**Additional Info Gathered:** {additional_info_from_step2} (if any)

Formulate your response:
1.  Acknowledge the issue.
2.  Provide a step-by-step solution, a link to relevant resources, or a clear explanation of the next steps the customer or you will take.
3.  End with an offer for further assistance.

If a direct solution isn't possible (e.g., requires live agent), state that and prepare for escalation.
```

### Step 4: Determine Escalation Rule

**Goal:** Decide if the issue requires escalation to a specialist or a different department, based on the issue complexity, urgency, and whether the proposed solution was sufficient.

**Prompt:**

```markdown
You are an internal process manager for customer support. Based on the customer's initial issue, its classification, and the proposed solution, determine if this issue requires escalation.

**Issue Category:** {issue_category}
**Urgency Level:** {urgency_level}
**Proposed Solution:** {proposed_solution_from_step3}

**Criteria for Escalation:**
-   'Critical' urgency level.
-   Issue category is 'Technical Support' and proposed solution was 'requires live agent'.
-   Issue category is 'Billing' and requires access to sensitive account information not available to a general agent.

Output 'YES' if escalation is required, along with a brief reason and suggested department (e.g., 'Technical Team', 'Billing Department'). Otherwise, output 'NO'.

Format your response as a JSON object with keys 'escalate' (boolean), 'reason' (string, if escalate is true), and 'department' (string, if escalate is true).
```

This chain creates a clear, multi-step flow where each stage builds upon the information gathered in the previous one, fulfilling the criteria for Exercise 1.
